```python
def romanToInt(s):
    
    symbol = {
        "I": 1,
        "V": 5,
        "X": 10,
        "L": 50,
        "C": 100,
        "D": 500,
        "M": 1000
    }

    total = 0
    
    for i in range(len(s)):
        num = symbol[s[i]]
        total = num + total
    return total
                
romanToInt('MCMXCIV')          
        
        
    
```
