```python
def mergeTwoLists(list1, list2):

    merge_list = []
    
    for i, j in zip(list1, list2):
        merge_list.append(i)
        merge_list.append(j)
        
    merge_list.sort()
    print(merge_list)


mergeTwoLists([1,2,4], [1,3,4])
```

```python
def compute_lists(list1, list2):

    computed_list = []
    
    for i, j in zip(list1, list2):
        product = i * j
           computed_list.append(product)

    return computed_list

compute_lists([1,2,4], [1,3,4])
```
