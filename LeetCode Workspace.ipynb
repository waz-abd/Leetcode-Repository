```python
def isValid(s):

    stack = []

    pairs = {
        ")":"(",
        "]":"[",
        "}":"{"
    }

    for ch in s:
        if ch not in pairs:
            stack.append(ch)
        else:
                
            if not stack:
                return False
                
            if stack[-1] != pairs[ch]:
                return False
                
            stack.pop()

    return not stack
            
isValid('([)')
```
